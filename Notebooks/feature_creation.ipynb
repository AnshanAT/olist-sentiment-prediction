##### Read in the master df

```python
import pandas as pd
import numpy as np

master_df = pd.read_csv("../Data/master_df.csv")

# drops the unnamed col... 
master_df = master_df.drop(columns=['Unnamed: 0'])

# print out a list of all the cols
print(list(master_df.columns))

master_df.head()
```

##### Bit of Cleaning

```python
# 
master_df['product_name_length']  = master_df['product_name_lenght']        
master_df['product_desc_length']  = master_df['product_description_lenght']


master_df.drop(columns=['product_description_lenght', 'product_name_lenght'])
```

##### Haversine Distance (distance between two points given long and lat)

```python
def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])

    a = np.sin((lat2-lat1)/2)**2 + np.cos(lat1)*np.cos(lat2)*np.sin((lon2-lon1)/2)**2

    return R * 2 * np.arcsin(np.sqrt(a))

master_df['geo_distance_km'] = haversine(
    master_df['seller_lat'], master_df['seller_lng'],
    master_df['customer_lat'], master_df['customer_lng']
)
```

##### Datetime cols convert to datetime

```python
date_time_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date',
    'shipping_limit_date',
    'review_creation_date',
    'review_answer_timestamp',
]
for col in date_time_cols:
    master_df[col] = pd.to_datetime(master_df[col], errors='coerce')
```

##### Feature Construction Time Related

86400 seconds in a day, will be taking the total seconds and dividing...

- delivery_delay_days
- 

```python
# delivery delay 
master_df['delivery_delay_days'] = (
    master_df['order_delivered_customer_date'] - master_df['order_estimated_delivery_date']
).dt.total_seconds() / (86400)

# how long delivery took from purchase 
master_df['end_to_end_delivery_days'] = (
    master_df['order_delivered_customer_date'] - master_df['order_purchase_timestamp']
).dt.total_seconds() / 86400

# Estimated delivery days
master_df['estimated_delivery_days'] = (
    master_df['order_estimated_delivery_date'] - master_df['order_purchase_timestamp']
).dt.total_seconds() / 86400

# Ratio of actual / estimated delivery days (pos means bad, neg means good) pos = longer to take than they said...
master_df['delivery_ratio'] = (
    master_df['end_to_end_delivery_days'] / master_df['estimated_delivery_days']
)

# Approval time... 
master_df['processing_time_hrs'] = (
    master_df['order_approved_at'] - master_df['order_purchase_timestamp']
).dt.total_seconds() / 3600


# --- 3h. days from delivery to review (how fast did they review?) ---
master_df['days_to_review'] = (
    master_df['review_creation_date'] - master_df['order_delivered_customer_date']
).dt.total_seconds() / 86400

# Calendar features ---
master_df['purchase_hour']  = master_df['order_purchase_timestamp'].dt.hour
master_df['purchase_dow']   = master_df['order_purchase_timestamp'].dt.dayofweek   
master_df['purchase_month'] = master_df['order_purchase_timestamp'].dt.month
master_df['is_weekend']     = (master_df['purchase_dow'] >= 5).astype(int)
master_df['is_night_purchase'] = master_df['purchase_hour'].between(22, 5).astype(int)  

# shipping limit proximity (did seller ship close to the deadline lower means closer to the deadline
master_df['days_before_ship_limit'] = (
    master_df['shipping_limit_date'] - master_df['order_delivered_carrier_date']
).dt.total_seconds() / 86400

```

##### Freight & pricing features

```python
# How much of the price is delivery...
master_df['freight_ratio'] = (
    master_df['freight_value'] / (master_df['price'] + master_df['freight_value'])
)

```

##### Product features

```python

# Has description / has photos flags 
master_df['has_description'] = (master_df['product_desc_length'] > 0).astype(int)
master_df['has_photos']      = (master_df['product_photos_qty'] > 0).astype(int)
master_df['photo_qty']       = master_df['product_photos_qty'].fillna(0)
```
